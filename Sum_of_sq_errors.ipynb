```python
import numpy as np 

def sum_of_sq_erros(predicted, actual):
    sum_error=0
    # here we shall receive the predicted one and we have the actual one already 
    # assuming that both the arrays are of same length 
    length=len(predicted)
    final=np.empty(length)
    for i in range(length):
        sub=abs(predicted[i]-actual[i])
        final[i]=sub
    print("FInal Array: ", final)
    # now the sum 
    for i in final:
        sum_error+=(i*i)
    print("Sum of error sq : ",sum_error)
    
predicted=np.array([2,4])
actual=np.array([1,2])
sum_of_sq_erros(predicted,actual)
```
