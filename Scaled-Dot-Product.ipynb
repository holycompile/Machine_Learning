# Scaled-Dot Product Basic Implementation

| Variable | Meaning | Purpose |
|---|---|---|
| Q | Query | What information am I looking for? |
| K | Key | What information do I contain? |
| V | Value | What information do I provide? |

Think of it like a search engine:
- Query: What am I searching for?
- Key: What does each result contain?
- Value: What information should I retrieve?

We will take the values of Query, Key and Value by ourselves for now 

```python
import numpy as np


# Query matrix
Q = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

# Key matrix
K = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

# Value matrix
V = np.array([
    [10, 0],
    [0, 10],
    [5, 5]
])
```

## 📐 The Complete Mathematical Pipeline


1. **Calculate Similarity (Raw Scores):**
   $$\text{Scores} = Q K^T$$
   *Measures how strongly each Query token aligns with every Key token.*

2. **Scale the Scores:**
   $$\text{Scaled} = \frac{\text{Scores}}{\sqrt{d_k}}$$
   *Scaling by $\frac{1}{\sqrt{d_k}}$ keeps variance stable as the vector dimension $d_k$ grows.*

3. **Apply Softmax Normalization:**
   $$\text{Weights} = \text{Softmax}(\text{Scaled})$$
   *Ensures each row forms a valid attention probability distribution (all weights $\ge 0$ and $\sum = 1$).*

4. **Multiply by Values:**
   $$\text{Output} = \text{Weights} \cdot V$$
   *Generates the final attention output as a convex combination of Value vectors.*

---

### 🎯 The Universal Transformer Attention Equation

$$\mathbf{Attention}(Q, K, V) = \text{Softmax}\left(\frac{Q K^T}{\sqrt{d_k}}\right) V$$


```python
# Malking the transpose of K 1st 

K_transposed=np.transpose(K)

print(K_transposed)

scores = np.dot(Q,K_transposed)

print(f"Scores:\n\n{scores}")

#------------------------------------
d_k=2 # for this particular program implementation

scaled = scores / np.sqrt(d_k)

print(f"Scaled matrix: \n \n {scaled}")
print("----")
```

### Softmax

The formula is:

$$Softmax(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$

**Important implementation detail:**  
Apply Softmax row-wise, and subtract the maximum value in each row before exponentiating. This makes the calculation numerically stable:

$$Softmax(x_i) = \frac{e^{x_i - \max(x)}}{\sum_j e^{x_j - \max(x)}}$$

```python
def softmax(scaled):
    rows_max=np.max(scaled, axis=1, keepdims=True)
    exp_values=np.exp(scaled-rows_max)
    attention_weights = exp_values / np.sum(exp_values, axis=1, keepdims=True)
    return attention_weights

weights=softmax(scaled)

print(f"Weights: \n\n {weights}")

# Final Output
output= np.dot(weights,V)

print("FInal Output is here : \n\n", output)
```
