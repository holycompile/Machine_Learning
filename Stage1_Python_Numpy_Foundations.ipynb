## Rank, Shape and Size Understanding 

```python
import numpy as np

x=np.array([ 
            [10,20,30,40],
            [50,60,70,80]
          ])

rank=x.ndim
shape=x.shape
size=x.size

print(f"Rank {rank}  Shape {shape}  size {size}")
```
