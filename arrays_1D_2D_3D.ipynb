# NumPy 1D, 2D, and 3D Array Examples

Below are three separate, standalone Python code examples demonstrating how to create and inspect 1D, 2D, and 3D arrays in NumPy.

---

## Code 1: 1D Array (Vector)

A 1-dimensional array represents a single sequence or list of values (e.g., daily temperatures).

```python
import numpy as np

# 1D Array: 5 elements
temperatures = np.array([28.5, 30.1, 29.0, 31.2, 27.8])

print("1D Array:\n", temperatures)
print("Shape:", temperatures.shape)       # (5,)
print("Dimensions:", temperatures.ndim)   # 1
```

---

## Code 2: 2D Array (Matrix)

A 2-dimensional array represents a grid with rows and columns (e.g., student test scores across subjects).

```python
import numpy as np

# 2D Array: 3 students (rows) x 4 subjects (columns)
scores = np.array([
    [85, 90, 78, 92],  # Student 0
    [88, 76, 95, 89],  # Student 1
    [91, 84, 82, 94]   # Student 2
])

print("2D Array:\n", scores)
print("Shape:", scores.shape)       # (3, 4)
print("Dimensions:", scores.ndim)   # 2
```

---

## Code 3: 3D Array (Tensor)

A 3-dimensional array represents a stack of 2D matrices (e.g., `(B, T, D)` data for 2 sensors over 3 time steps with 2 features each).

```python
import numpy as np

# 3D Array: 2 sensors (batch) x 3 time steps x 2 features
sensor_data = np.array([
    # Sensor 0
    [
        [22.1, 55.0],  # Time 0: [Temp, Humidity]
        [22.5, 54.2],  # Time 1
        [23.0, 53.8]   # Time 2
    ],
    # Sensor 1
    [
        [19.4, 68.1],  # Time 0: [Temp, Humidity]
        [19.8, 67.5],  # Time 1
        [20.2, 66.0]   # Time 2
    ]
])

print("3D Array:\n", sensor_data)
print("Shape:", sensor_data.shape)       # (2, 3, 2)
print("Dimensions:", sensor_data.ndim)   # 3
```
